```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("hotel_sales.csv")

df["Booking_Date"] = pd.to_datetime(df["Booking_Date"])
df["Check_In"] = pd.to_datetime(df["Check_In"])
df["Check_Out"] = pd.to_datetime(df["Check_Out"])
```

```python
monthly_revenue = df.groupby(
    df["Booking_Date"].dt.month
)["Total_Revenue"].sum()

monthly_revenue 
```

```python
monthly_revenue.plot(kind="line",marker="o")

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue (JYP)")
plt.xticks(range(1,13))
plt.grid(True)
plt.show()
```

```python
month_names = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
               "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

monthly_revenue.plot(kind="line",marker="o")

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue (JYP)")
plt.xticks(range(1,13),month_names)
plt.grid(True)
plt.show()
```

```python
highest_month = monthly_revenue.idxmax()
highest_revenue = monthly_revenue.max()

lowest_month = monthly_revenue.idxmin()
lowest_revenue = monthly_revenue.min()

print("Highest month:", highest_month)
print("Highest revenue:", highest_revenue)

print("Lowest month:", lowest_month)
print("Lowest revenue:", lowest_revenue)
```

```python
revenue_by_city = df.groupby("City")["Total_Revenue"].sum().sort_values(ascending=False)
revenue_by_city
```

```python
revenue_by_city.plot(kind="bar")

plt.title("Total Revenue by City")
plt.xlabel ("City")
plt.ylabel("Total Revenue (JPY)")
plt.xticks(rotation=45)
plt.show()
```

```python
revenue_by_customer_channel = df.groupby(
    ["Customer_Type", "Booking_Channel"]
)["Total_Revenue"].sum().unstack()

revenue_by_customer_channel
```

```python
revenue_by_customer_channel.plot(kind="bar")

plt.title("Revenue by Customer Type and Booking Channel")
plt.xlabel("Customer Type")
plt.ylabel("Total Revenue (JPY)")
plt.xticks(rotation=0)
plt.legend(title="Booking Channel")
plt.show()
```
